```python
import numpy as np

# Actual marks
actual = np.array([55,58,77,76])

# Predicted marks
predicted = np.array([50,60,70,80])

# Residuals
residuals = actual - predicted

print("Residuals:", residuals)

# Squared residuals
squared = residuals**2

print("Squared residuals:", squared)

# Sum of squared residuals
print("SSR:", np.sum(squared))
```

```python
import numpy as np 
# Actual values 
actual = np.array([50,60,70,80])
# Predicted values 
predicted = np.array([52,61,69,79])
#Residuals 
residuals =actual- predicted 
#RSS
rss=np.sum(residuals **2) 
#mean of actual values 
mean_actual = np.mean(actual) 
#TSS
tss =np.sum((actual- mean_actual)**2)
#R2 squared 
r2 = 1-(rss/tss)
print("RSS:",rss)
print ("TSS:",tss)
print ("R2:",r2)
```

```python
import numpy as np
import statsmodels.api as sm
from statsmodels.stats.diagnostic import acorr_breusch_godfrey

# Create X values
X = np.arange(1, 101)

# Create Y values
Y = 5 + 2*X + np.random.normal(0,10,100)

# Add constant for intercept
X = sm.add_constant(X)

# Run OLS regression
model = sm.OLS(Y, X).fit()

# Breusch-Godfrey test
bg_test = acorr_breusch_godfrey(model, nlags=1)

# Print results
print("LM Statistic:", bg_test[0])
print("P-value:", bg_test[1])
```

```python
#example of spurious regression 
import numpy as np 
import statsmodels.api as sm 
import matplotlib.pyplot as plt 

#fixes randomness 
np.random.seed(42)

#create two unrelated random walks 
x = np.cumsum(np.random.normal(0,1,100))
y = np.cumsum(np.random.normal(0,1,100))

#add cosntant 
x_const = sm.add_constant(x)

#run regression 
model = sm.OLS(y,x_const).fit()

#print summary 
print(model.summary())
```

```python
from statsmodels.tsa.stattools import adfuller

result_x = adfuller(x)
result_y = adfuller(y)

print("X p-value:", result_x[1])
print("Y p-value:", result_y[1])
```

```python
import numpy as np
from statsmodels.tsa.stattools import adfuller

# Fix randomness
np.random.seed(42)

# Create random walk
errors = np.random.normal(0,1,100)
random_walk = np.cumsum(errors)

# Difference the series
diff_series = np.diff(random_walk)

# ADF test again
result = adfuller(diff_series)

# Print result
print("P-value:", result[1])
```

```python
import numpy as np
import matplotlib.pyplot as plt 
# fix randomness 
np.random.seed(42)
#number of observations 
n=100
#white noise shocks 
errors = np.random.normal(0,1,n)
#empty series 
y = np.zeros(n)
#AR (1)  coefficient 
phi = 0.8
#generate ar(1)
for t in range (1,n):
    y[t] = phi*y[t-1]+errors[t]
#plot 
plt.plot(y)
plt.title("AR(1) process")
plt.xlabel("Time")
plt.ylabel("value")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# fix randomness
np.random.seed(42)

# number of observations
n = 100

# white noise shocks
errors = np.random.normal(0,1,n)

# empty series
y = np.zeros(n)

# coefficients
phi = 0.8
theta = 0.5

# generate ARMA(1,1)
for t in range(1,n):
    y[t] = phi*y[t-1] + errors[t] + theta*errors[t-1]
```

```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
import matplotlib.pyplot as plt

plot_acf(y, lags=20)
plt.title("ACF of ARMA(1,1)")
plt.show()

plot_pacf(y, lags=20)
plt.title("PACF of ARMA(1,1)")
plt.show()
```

```python
# creating a random walk for ARIMA 
import numpy as np 
import matplotlib.pyplot as plt 

# fix randomness 
np.random.seed(42)
#number of observation 
n= 100
#white noise 
errors =np.random.normal(0,1,n)
#random walk 
random_walk = np.cumsum(errors)
#plotiing 
plt.plot(random_walk)
plt.title ("Random walk (non stationary)")
plt.xlabel("time")
plt.ylabel("value")
plt.show 
```

```python
diff_series = np.diff(random_walk)

plt.plot(diff_series)
plt.title("Differenced Random Walk")
plt.xlabel("Time")
plt.ylabel("Value")
plt.show()
```

```python
from statsmodels.tsa.stattools import adfuller 
result = adfuller(diff_series)
print ("ADF Statistics:", result[0])
print ("p-value:",result[1])
```
